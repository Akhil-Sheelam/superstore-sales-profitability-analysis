```python
# Group the data by Sub-Category and calculate total Sales and total Profit
# This helps us identify which product sub-categories are contributing
# the most and least to overall profitability.

subcategory_profit = (
    df.groupby("Sub-Category")[["Sales", "Profit"]]
      .sum()
      .sort_values("Profit")
      .round(2)
)

# Display the result
subcategory_profit
```

```python
# Calculate profit margin for each sub-category.
# Profit Margin = (Profit / Sales) × 100
# This tells us how much profit the business makes for every ₹/$100 of sales.

subcategory_profit["Profit Margin %"] = (
    subcategory_profit["Profit"] /
    subcategory_profit["Sales"] * 100
).round(2)

# Display Sales, Profit, and Profit Margin together
subcategory_profit
```

```python
# Select only Tables and Bookcases.
# These two sub-categories have negative overall profit,
# so we want to investigate their discount levels.

furniture_loss = df[
    df["Sub-Category"].isin(["Tables", "Bookcases"])
]

# Calculate total sales, total profit, average discount,
# and maximum discount for each sub-category.

furniture_discount = furniture_loss.groupby("Sub-Category").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Average_Discount=("Discount", "mean"),
    Maximum_Discount=("Discount", "max")
).round(2)

# Display the result
furniture_discount
```

```python
# Analyze profitability at each discount level.
# This helps us see whether higher discounts are associated
# with lower profit for Tables and Bookcases.

discount_profit = (
    furniture_loss
    .groupby(["Sub-Category", "Discount"])["Profit"]
    .agg(
        Transactions="count",
        Total_Profit="sum",
        Average_Profit="mean"
    )
    .round(2)
)

# Display the result
discount_profit
```

```python
# Filter the dataset to Tables and Bookcases.
# We want to investigate where these loss-making products
# are generating negative profit.

furniture_loss = df[
    df["Sub-Category"].isin(["Tables", "Bookcases"])
]

# Group the data by Region and Sub-Category.
# Calculate total Sales and total Profit for each combination.

region_loss = (
    furniture_loss
    .groupby(["Region", "Sub-Category"])[["Sales", "Profit"]]
    .sum()
    .round(2)
)

# Display the regional profitability
region_loss
```

```python
# Calculate Profit Margin for each Region and Sub-Category.
# This helps us compare profitability across regions
# even when sales volumes are different.

region_loss["Profit Margin %"] = (
    region_loss["Profit"] /
    region_loss["Sales"] * 100
).round(2)

# Display the final result
region_loss
```

```python
# Filter the dataset to only Table products.
# We are investigating which individual products
# contribute most to the overall Table loss.

tables = df[
    df["Sub-Category"] == "Tables"
]

# Group by Product Name and calculate:
# 1. Total Sales
# 2. Total Profit
# 3. Average Discount
# 4. Number of transactions

table_products = (
    tables
    .groupby("Product Name")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Average_Discount=("Discount", "mean"),
        Transactions=("Product Name", "count")
    )
    .sort_values("Profit")
    .round(2)
)

# Display the products from the biggest losses
# to the highest profits.
table_products
```

```python
# Select only products where total profit is negative.
# These products are contributing to the overall Table loss.

loss_making_tables = table_products[
    table_products["Profit"] < 0
]

# Display the loss-making Table products
loss_making_tables
```

```python
# Select the 10 Table products with the largest total losses.
# Sorting by Profit in ascending order puts the biggest losses first.

top_10_table_losses = (
    table_products
    .sort_values("Profit", ascending=True)
    .head(10)
)

# Display the top 10 loss-making Table products
top_10_table_losses
```

```python
# Calculate the total loss across all Table products.

total_table_loss = table_products["Profit"].sum()

# Calculate the combined profit/loss of the 10 worst-performing products.

top_10_loss = top_10_table_losses["Profit"].sum()

# Calculate what percentage of the total Table loss
# comes from these 10 products.

loss_contribution = (top_10_loss / total_table_loss * 100).round(2)

print("Total Table Profit/Loss:", round(total_table_loss, 2))
print("Top 10 Table Profit/Loss:", round(top_10_loss, 2))
print("Top 10 Loss Contribution (%):", loss_contribution)
```

```python
# Create the Year column from Order Date.
# This will allow us to analyze yearly performance in Power BI.

df["Year"] = df["Order Date"].dt.year


# Create the Month column.
# This extracts the month number from Order Date.

df["Month"] = df["Order Date"].dt.month


# Create a Month-Year column.
# This is useful for displaying monthly trends in Power BI.

df["Month-Year"] = df["Order Date"].dt.to_period("M").astype(str)


# Create Profit Margin.
# Profit Margin = Profit / Sales × 100

df["Profit Margin %"] = (
    df["Profit"] / df["Sales"] * 100
).round(2)


# Create a Profit/Loss Status column.
# If Profit is greater than or equal to zero,
# classify the transaction as Profitable.
# Otherwise, classify it as Loss.

df["Profit/Loss Status"] = df["Profit"].apply(
    lambda x: "Profit" if x >= 0 else "Loss"
)


# Display the first 5 rows to verify the new columns.

df.head()
```

```python
# Create groups for different discount levels.
# This makes it easier to analyze the relationship
# between discount levels and profitability.

def classify_discount(discount):
    
    # No discount
    if discount == 0:
        return "0%"
    
    # Low discount
    elif discount <= 0.20:
        return "1-20%"
    
    # Medium discount
    elif discount <= 0.40:
        return "21-40%"
    
    # High discount
    else:
        return "40%+"

    
# Apply the function to the Discount column.

df["Discount Band"] = df["Discount"].apply(classify_discount)


# Check how many records fall into each discount band.

df["Discount Band"].value_counts()
```
